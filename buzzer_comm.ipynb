```python
from pynq.overlays.base import BaseOverlay
from pynq.lib import Pmod_PWM
import time

# Load the base overlay
base = BaseOverlay("base.bit")

# Set up PWM on PMODA pin 3
buzzer = Pmod_PWM(base.PMODA, 3)

# Access onboard buttons
btns_gpio = base.buttons

def play_tone(frequency=1000, duration=0.5):
    """Generate a short beep using PWM."""
    period_ms = int(1000000 / frequency)  # Convert Hz to period in µs
    duty_cycle = 20  # Lower duty cycle to avoid continuous ringing

    buzzer.generate(period_ms, duty_cycle)  # Start PWM
    time.sleep(duration)  # Keep tone active
    buzzer.stop()  # Stop PWM to prevent continuous ringing

# Main loop to listen for button presses
while True:
    btn_state = btns_gpio.read()
    
    if btn_state & 0b0001:  # BTN0 - Connect
        print("Connected")
    
    if btn_state & 0b0010:  # BTN1 - Beep
        print("Beeping!")
        play_tone(1000, 0.5)  # Play 1kHz tone for 0.5s
    
    if btn_state & 0b0100:  # BTN2 - Disconnect
        print("Disconnected")
        break  # Exit loop
```
